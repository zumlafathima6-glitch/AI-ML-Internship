```python

# ==========================================
# STUDENT PERFORMANCE ANALYSIS SYSTEM
# Using Python
# ==========================================

# Subject names
subjects = ["Mathematics", "Science", "English"]

# Store all student records
students = []


# Function to calculate grade
def calculate_grade(average):

    if average >= 75:
        return "A"

    elif average >= 65:
        return "B"

    elif average >= 50:
        return "C"

    elif average >= 35:
        return "S"

    else:
        return "F"


# Function to calculate student performance
def analyze_student(name, marks):

    total = sum(marks)
    average = total / len(marks)

    highest = max(marks)
    lowest = min(marks)

    grade = calculate_grade(average)

    return {
        "name": name,
        "marks": marks,
        "total": total,
        "average": average,
        "highest": highest,
        "lowest": lowest,
        "grade": grade
    }


# Function to display individual report
def display_report(student):

    print("\n" + "=" * 40)
    print("       STUDENT PERFORMANCE REPORT")
    print("=" * 40)

    print("Student Name :", student["name"])

    for i in range(len(subjects)):
        print(subjects[i], ":", student["marks"][i])

    print("-" * 40)

    print("Total Marks  :", student["total"])
    print("Average      :", round(student["average"], 2))
    print("Highest Mark :", student["highest"])
    print("Lowest Mark  :", student["lowest"])
    print("Grade        :", student["grade"])

    print("=" * 40)


# Function to display class analysis
def class_analysis():

    print("\n" + "=" * 45)
    print("          CLASS PERFORMANCE ANALYSIS")
    print("=" * 45)

    averages = [student["average"] for student in students]

    class_average = sum(averages) / len(averages)

    topper = max(students, key=lambda s: s["average"])

    lowest_student = min(
        students,
        key=lambda s: s["average"]
    )

    print("Total Students :", len(students))

    print("Class Average  :", round(class_average, 2))

    print("Class Topper   :", topper["name"])

    print("Highest Average:", round(topper["average"], 2))

    print("Lowest Student :", lowest_student["name"])

    print("Lowest Average :", round(lowest_student["average"], 2))

    # Grade distribution
    print("\nGRADE DISTRIBUTION")

    for grade in ["A", "B", "C", "S", "F"]:

        count = sum(
            1 for student in students
            if student["grade"] == grade
        )

        print(grade, ":", count, "students")


# ==========================================
# MAIN PROGRAM
# ==========================================

print("=" * 45)
print("   STUDENT PERFORMANCE ANALYSIS SYSTEM")
print("=" * 45)

number_of_students = int(
    input("\nEnter number of students: ")
)

# Get student information
for i in range(number_of_students):

    print("\nEnter details for Student", i + 1)

    name = input("Enter student name: ")

    marks = []

    for subject in subjects:

        while True:

            try:
                mark = float(
                    input(f"Enter {subject} marks (0-100): ")
                )

                if 0 <= mark <= 100:
                    marks.append(mark)
                    break

                else:
                    print("Please enter marks between 0 and 100.")

            except ValueError:
                print("Invalid input! Enter a numerical value.")

    # Analyze student
    student = analyze_student(name, marks)

    students.append(student)


# Display all student reports
for student in students:

    display_report(student)


# Display overall class analysis
class_analysis()

print("\nAnalysis completed successfully!")
```
